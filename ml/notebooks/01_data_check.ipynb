```python
import pandas as pd
import numpy as np

# 1. Load the Excel file with the correct header row
file_path = "../data/raw/Projects_Report.xlsx"

df = pd.read_excel(file_path, header=2)

# 2. Clean the column names
df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

# 3. Rename columns
df = df.rename(columns={
    "Sr. No.": "sr_no",
    "Sector Name": "sector",
    "Line Ministry": "ministry",
    "Implementing Agency": "agency",
    "Project Code": "project_code",
    "Project Name": "project_name",
    "Original Cost (in cr.)": "original_cost",
    "Revised Cost (in cr.)": "revised_cost",
    "Expenditure (in cr.)": "expenditure",
    "Physical Progress (in %)": "physical_progress",
    "Original Date of Commissioning": "original_completion",
    "Revised Date of Commissioning": "revised_completion",
    "Sanction Date": "sanction_date"
})

# 4. Convert numeric columns
numeric_cols = [
    "original_cost",
    "revised_cost",
    "expenditure",
    "physical_progress"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 5. Convert date columns
date_cols = [
    "original_completion",
    "revised_completion",
    "sanction_date"
]

for col in date_cols:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce",
        dayfirst=True
    )

# 6. Remove completely empty rows
df = df.dropna(how="all")

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```
